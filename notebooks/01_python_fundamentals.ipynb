```python
vibration_readings = [4.2, 7.2, 5.1, 8.3, 3.9]

print(vibration_readings)
print("Number of readings:", len(vibration_readings))
```

```python
for reading in vibration_readings:
    if reading >6:
        print("High vibration warning:", reading)
    else:
        print("Vibration is within the configured range:", reading)
```

```python
machine = {
    "machine_id": "M101",
    "temperature": 78.5,
    "vibration": 4.2,
    "operating_hours": 3200
}

print(machine)
print("Machine ID:", machine["machine_id"])
print("Temperature:", machine["temperature"])
print("Vibration:", machine["vibration"])
```

```python
print("Machine_operating_hours:",machine["operating_hours"])
machine["vibration"] = 7.2
print("Updated vibration:", machine["vibration"])
if machine["vibration"] > 6:
    print("High vibration warning:", machine["vibration"])
```

```python
def check_vibration(vibration):
    if vibration > 6:
        print("High vibration warning:", vibration)
    else:
        print("Vibration is within the configured range:", vibration)

check_vibration(4.2)
check_vibration(7.2)
check_vibration(5.1)
```

```python
def check_temperature(temperature):
    if temperature > 80:
        print("High temperature warning:", temperature)
    else:
        print("Temperature is within the configured range:", temperature)

check_temperature(75)
check_temperature(85)
check_temperature(80)
```
